```python
import pandas as pd

# 定义一个转换函数：将数值转换为6位字符串，前面补0
def to_six_digit_str(x):
    try:
        return f"{int(x):06d}"
    except:
        return x  # 如果转换失败，保留原值（可选）

# 读取CSV，并对 'Stkcd' 列应用转换函数
df_inner = pd.read_csv(
    'inner.csv',
    dtype={'company_id': str},  # 先以字符串读入（防止前置0被丢弃）
    converters={'company_id': to_six_digit_str}  # 再用函数处理
)

df_st = pd.read_csv(
    'fdp.csv',
    dtype={'company_id': str},  # 先以字符串读入（防止前置0被丢弃）
    converters={'company_id': to_six_digit_str}  # 再用函数处理
)
```
